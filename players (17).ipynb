# Classification of skills

```python
# 1-attacking_score
attacking_cols = [
    'finishing', 'heading_accuracy', 'volleys',
    'shot_power', 'long_shots', 'positioning', 'penalties'
]

# 2-playmaking_score
playmaking_cols = [
    'crossing', 'short_passing', 'long_passing',
    'dribbling', 'ball_control', 'curve', 'freekick_accuracy', 'vision', 'composure'
]

# 3-athletic_score
athletic_cols = [
    'acceleration', 'sprint_speed', 'agility',
    'reactions', 'balance', 'jumping', 'stamina', 'strength'
]

# 4-defending_score
defending_cols = [
    'marking', 'standing_tackle', 'sliding_tackle',
    'interceptions', 'aggression'
]
```

```python
# Average of each category for each player

df['attacking_score'] = df[attacking_cols].mean(axis=1).round(1)
df['playmaking_score'] = df[playmaking_cols].mean(axis=1).round(1)
df['athletic_score'] = df[athletic_cols].mean(axis=1).round(1)
df['defending_score'] = df[defending_cols].mean(axis=1).round(1)
```

# Save the skills in a file and then delete them
We saved the raw skill data before categorizing them in a separate file so we could get them back if we needed, and then we deleted it.

```python
# save the raw skill data before categorizing them in a separate file

all_skill_cols = attacking_cols + playmaking_cols + athletic_cols + defending_cols

df[['name'] + all_skill_cols].to_csv('raw_skills_data.csv', index=False)
```

```python
# Delete the skills after categorizing them and save them in another file

df = df.drop(columns=all_skill_cols)
```

# handling outliers

We need to check if this value is an input error or an actual reality



```python
IQR = df['value_euro'].quantile(0.75) - df['value_euro'].quantile(0.25)
upper_bound = df['value_euro'].quantile(0.75) + 1.5 * IQR

outliers_mask = df['value_euro'] > upper_bound

print('IQR: ', IQR)
print('Upper bound: ', upper_bound)
print('Number of outliers: ', outliers_mask.sum())
```

Outlier Analysis:

    Using the IQR method, 2,069 players were identified as outliers above the upper bound of €5,187,500. These observations were retained because extremely high market values can represent legitimate high-value players rather than data errors. Therefore, removing them could lead to losing meaningful information about the upper end of the football transfer market.

# features engineering

```python
# 1-log_value
# This column makes the relationship with the evaluation almost linear and easy to understand for the chart and the correlation coefficient.

df['log_value_euro'] = np.log1p(df['value_euro'])

df[['name', 'value_euro', 'log_value_euro']].head()
```

```python
# 2-growth_potential
# It explains the reason for paying huge amounts for talents whose current evaluation is average.

df['growth_potential'] = df['potential'] - df['overall_rating']

df[['name', 'overall_rating', 'potential', 'growth_potential']].head(10)

df['growth_potential'].value_counts()
```

```python
# 3-age_group
# To understand the effect of price drop as you age even if the valuation is high

bins = [0, 22, 29, 34, float('inf')]
labels = ['Young Talent (<22)', 'Peak (22-28)', 'Experienced (29-33)', 'Veteran (34+)']

df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=False)

df[['name', 'age', 'age_group']].head()

df.groupby('age_group', observed=False)['age'].agg(
    ['min', 'max', 'count']
)
```

# Q1: The relationship between overall rating and market value.

    Does the player with the highest overall_rating necessarily have the highest value_euro?

    Is the relationship linear, or are there obvious differences?

```python
# 1- Analysis of the market value distribution of players and Analysis of overall_rating distribution

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(df['log_value_euro'], bins=20, color='blue', edgecolor='black')
axes[0].set_title('Distribution of Value in Euros')
axes[0].set_xlabel('Value in Euros',color='blue')
axes[0].set_ylabel('Players',color='blue')


axes[1].hist(df['overall_rating'], bins=20, color='lightcoral', edgecolor='black')
axes[1].set_title('Distribution of Overall Rating')
axes[1].set_xlabel('Overall Rating', color= 'lightcoral')
axes[1].set_ylabel('Players', color= 'lightcoral')

plt.tight_layout()
plt.show()
```

1- overall rating distribution:

    It seems that the data roughly follows a normal distribution, with most players concentrated in the middle range (between 60 and 70), while the number of players gradually decreases as we move towards very low or very high ratings (big stars above 85).


2- market value distribution:


     It shows the distribution of player values (after applying the logarithmic transformation log_value_euro likely, since the values range between 10 and 18 on the axis).

```python
# 2- Does the player with the highest overall_rating necessarily have the highest value_euro?

plt.figure(figsize=(8, 5))

plt.scatter(
    df['overall_rating'],
    df['log_value_euro'],
    alpha=0.4,
    edgecolors="black"
)

plt.title('Overall Rating vs Market Value')
plt.xlabel('Overall Rating')
plt.ylabel('log Market Value (€)')

plt.show()
```

```python
# Proving the relationship with statistics using the correlation coefficient

df[['overall_rating', 'log_value_euro']].corr()
```

Summary of the first question:

  At first, we analyzed the distribution of market value and overall rating.
    
    
  Secondly, we analyzed the relationship between market value and overall evaluation and found the following:



    1.  At the beginning, we used the logarithm of market value to avoid extreme values.

    2.  We found that there is a positive relationship between market value and overall rating: the higher the overall rating, the higher the price.

    3. We also found that most players gather in the middle, which is the area that contains values from 60-80. There are also some players below or above that, but the main concentration of players is in the middle region.

    4. The correlation coefficient is approximately 0.949, indicating a very strong positive relationship between overall rating and log market value.


# Q2: The impact of different skills on market value.

```python
# The relationship of each type of skill with market value

skills = [
    'attacking_score',
    'playmaking_score',
    'athletic_score',
    'defending_score'
]

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

for ax, skill in zip(axes.flat, skills):

    ax.scatter(
        df[skill],
        df['log_value_euro'],
        alpha=0.3,
        edgecolors="black"
    )

    ax.set_title(f'{skill.replace("_", " ").title()} vs log Market Value')
    ax.set_xlabel(skill.replace("_", " ").title())
    ax.set_ylabel('log Market Value (€)')

plt.suptitle(
    'Player Skill Groups vs log Market Value',
    fontsize=16
)

plt.tight_layout()
plt.show()
```

This panel of four combined charts (subplots) shows the analysis of the second research question (RQ2), which looks at the relationship between player skill group indicators and log market value:

  1. Attacking skills (Attacking Score vs log Market Value - top left):



    * The diagram shows a small number in the area between 20-30, then it gradually increases until it reaches 85, where there is a big crowd among players, then it starts to decrease in the range between 85 to 90 and above.

    * The point distribution takes an upward shape, showing that the maximum values increase as the offensive skill improves.


  2. Playmaking skills (Playmaking Score vs Market Value - top right):



    * The chart shows a noticeable number of players in the 30-40 range, then it starts to increase and cluster between 40-90, and then it decreases a lot, becoming almost nonexistent above 90.
    

  3. Physical Skills (Athletic Score vs Market Value - bottom left):


    *   It shows the relationship between physical indicators (speed, endurance, strength) and market value. Fitness remains an essential and required factor, but it often supports the player's core technical skills.
    
    *   The chart shows a small number of players, around 40-50, then it starts to increase and crowd until it reaches 85, and then it decreases again.

  4. Defense Skills (Defending Score vs Market Value - bottom right):




    *  The chart shows players clustering in the 20-80 area, then it starts to decrease between 80-90, and then it becomes almost nonexistent above 90.
    




-------------------------------------------


```python
# Proving the relationship with statistics using the correlation coefficient

corr =  df[['attacking_score','playmaking_score','athletic_score','defending_score', 'log_value_euro']].corr()

# 1-Attacking_score: 0.577
# 2-Playmaking_score: 0.679
# 3-Athletic_score: 0.560
# 4-defending_score: 0.238
```

    3-Athletic_score: 0.560 --> Moderate positive relationship
    4-defending_score: 0.238 --> Weak positive relationship

3- Key takeaways:

    1- Playmaking has the strongest positive correlation with logarithmic market value among the four skill groups.
    2- Attacking and Athletic: both have a moderate and relatively similar correlation.
    3- Defending: the weakest link to market value, but that doesn't mean defending isn't important for a player.

# Q4: The relationship between young players' growth potential and price

```python
# Filter only young players whose ages are less than 22

young_players = df[df['age_group'] == 'Young Talent (<22)']

```

```python
plt.figure(figsize=(10,8))

plt.scatter(
    young_players['growth_potential'],
    young_players['log_value_euro'],
    alpha=0.4,
    edgecolors="black"
)

plt.title('Growth Potential vs log Market Value')
plt.xlabel('Growth Potential')
plt.ylabel('log Market Value (€)')

plt.show()
```

Explaining the chart:

    1- When growth potential is low (around 5-8), there are points that reach up to 18 on the Y-axis (a very high value), and there are other points around 11 (a low value)

    2- With a high growth potential (20+), almost all the points are clustered between 11 and 14



```python
young_players[['growth_potential', 'log_value_euro']].corr()
```

Summary of Question Four:

    1- When growth_potential is low (around 5-8), there are points that reach up to 18 on the Y-axis (a very high value), and there are other points around 11 (a low value) — meaning there's a big variation in value at low potential.

    2- With a high growth potential (20+), almost all the points are clustered between 11 and 14 — meaning there's not a single player with really high growth who reached a value like 17 or 18.

    3- It means: the highest market values (18, 17, 16.5) are all found at low to medium growth potential (5-12), not at high potential.

    4- There is a weak-to-moderate negative correlation between growth potential and current market value (r = -0.269057), suggesting that players with higher room for improvement tend to have lower current market values.

    * The current rating (overall_rating) is the real strongest factor, and this confirms the Q1 result (90% R²) from another angle

# The final summary of the project

   

This project explored the main factors associated with football players' market value using an exploratory data analysis approach.

The analysis started with the FIFA players dataset and went through a complete data preparation process, including removing irrelevant variables, handling missing market values, checking duplicates, excluding goalkeepers from the skill-based analysis, engineering aggregated skill scores, transforming market value using a logarithmic scale, and creating additional features such as `growth_potential` and `age_group`.

After data preparation and validation, the final dataset contained **15,675 outfield players across 20 variables**.

The analysis focused on four main research questions:

1. **Does a higher overall rating necessarily mean a higher market value?**
2. **Which skill groups have the strongest relationship with market value?**
3. **How does age affect a player's market value?**
4. **How does growth potential relate to the current market value of young players?**

### Key Findings

- **Overall rating has a very strong positive relationship with market value**, with a correlation of approximately **0.949** with logarithmic market value.
- Among the four aggregated skill groups, **Playmaking** showed the strongest positive correlation with logarithmic market value (**0.679**).
- **Attacking** and **Athletic** skills showed moderate positive relationships with market value (**0.577** and **0.560**, respectively).
- **Defending** showed the weakest positive relationship among the four skill groups (**0.238**), although this does not mean defensive ability is unimportant in football.
- Market value generally increased from the **Young Talent** stage through the **Peak** and **Experienced** stages, before declining in the **Veteran** stage.
- For young players, higher `growth_potential` did not correspond to higher current market value. The analysis showed a weak-to-moderate negative correlation (**-0.269**), suggesting that players with greater room for improvement tend to have lower current market values at the present time.

### Overall Conclusion

The analysis suggests that **current player quality, represented by overall rating, has a much stronger relationship with current market value than potential for future improvement**.

Skill composition also matters, with **playmaking showing the strongest relationship among the analyzed skill groups**. However, market value is not determined by a single factor. Age, current ability, skill profile, and future potential can interact to produce very different valuations between players.

This project demonstrates a complete exploratory data analysis workflow, from data preparation and feature engineering to visualization, statistical analysis, and evidence-based conclusions.
